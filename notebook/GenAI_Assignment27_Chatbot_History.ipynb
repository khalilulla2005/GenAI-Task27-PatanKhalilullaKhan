## Assignment Overview
This assignment focuses on building stateful chatbots by managing conversation history in LangChain.


```
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain_community.chat_message_histories import ChatMessageHistory

llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0.7)
print("Environment initialized.")
```

## PART 1 — Messages & Message Placeholders
### Task 1: Conceptual Questions
1. **SystemMessage**: Sets the behavior/personality of the AI.
2. **HumanMessage**: Represents input from the user.
3. **AIMessage**: Represents the response generated by the AI.
4. **Message-based prompting**: Provides context, allowing the LLM to understand the conversation flow rather than treating each prompt as an isolated event.

### Task 2: MessagePlaceholder Usage


```
# Chat history list
history = ChatMessageHistory()
history.add_user_message("Hello, I am Patan.")
history.add_ai_message("Hello Patan! How can I help you?")

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])

chain = prompt | llm

print("=== Task 2: Injecting History ===")
print(chain.invoke({"history": history.messages, "input": "What is my name?"}).content)
```

## PART 2 — Conversation History Management
### Task 3 & 4: Storing and Trimming Chat History


```
# Task 4: Trimming
def trim_history(messages, max_messages=4):
    return messages[-max_messages:]

# Simulate long history
long_history = [HumanMessage(content=f"Msg {i}") for i in range(10)]
trimmed = trim_history(long_history)

print("=== Task 4: Trimming Results ===")
print(f"Original length: {len(long_history)}, Trimmed length: {len(trimmed)}")
print(f"Last 4 messages: {trimmed}")
```

## PART 3 — Stateful Q&A Chatbot
### Task 5 & 6: Building a Stateful Chatbot


```
# Task 6: Stateful Chatbot
class StatefulChatbot:
    def __init__(self, llm):
        self.llm = llm
        self.history = ChatMessageHistory()

    def chat(self, user_input):
        self.history.add_user_message(user_input)
        response = self.llm.invoke(self.history.messages)
        self.history.add_ai_message(response.content)
        return response.content

bot = StatefulChatbot(llm)
print("=== Task 6: Stateful Chat ===")
print("Bot:", bot.chat("I like Python."))
print("Bot:", bot.chat("What did I just say I like?"))
```

## Task 7: Observations & Insights
1. **Why chat history is important:** Enables context-aware responses, essential for conversational AI.
2. **Trade-offs:** Long history increases token usage and latency; trimming manages costs/limits but loses older context.
3. **Summarize vs Trim:** Summarize when you need to retain high-level context over long sessions; trim when recent messages are sufficient.
4. **Placeholder vs Memory:** Placeholder is the template mechanism to inject history; Memory is the component that stores/retrieves that history.
